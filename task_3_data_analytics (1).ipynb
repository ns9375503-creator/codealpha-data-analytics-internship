```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# --- TASK 3: DATA VISUALIZATION ---
np.random.seed(42)
study_hours = np.random.uniform(2, 12, 200)
exam_score = 45 + (4.5 * study_hours) + np.random.normal(0, 4, 200)

df_task3 = pd.DataFrame({
    'Study_Hours': study_hours,
    'Final_Exam_Score': exam_score
})

sns.set_theme(style="whitegrid")
plt.figure(figsize=(9, 5))

# Scatter plot with regression line
sns.regplot(
    x='Study_Hours',
    y='Final_Exam_Score',
    data=df_task3,
    color='teal',
    scatter_kws={'alpha': 0.6},
    line_kws={'color': 'darkorange', 'linewidth': 2.5}
)

plt.title('Task 3: Relationship Between Study Hours and Final Exam Scores', fontsize=13, fontweight='bold', pad=15)
plt.xlabel('Study Hours', fontsize=11)
plt.ylabel('Final Exam Score', fontsize=11)

plt.show()
```
