```python
import pandas as pd
import numpy as np

# Creating a clean Student Performance dataset for Task 2 EDA
np.random.seed(42)
data = {
    'Student_ID': range(1, 251),
    'Study_Hours': np.random.randint(1, 15, 250),
    'Attendance_Percentage': np.random.randint(50, 100, 250),
    'Previous_Exam_Score': np.random.randint(40, 95, 250),
    'Final_Exam_Score': np.random.randint(45, 100, 250)
}
df = pd.DataFrame(data)

print("--- 1. DATA STRUCTURE (Variables & Data Types) ---")
print(df.info())

print("\n--- 2. FIRST 5 ROWS (Data Preview) ---")
print(df.head())

print("\n--- 3. CHECK FOR MISSING VALUES (Data Issues) ---")
print(df.isnull().sum())

print("\n--- 4. STATISTICAL SUMMARY (Validating Assumptions) ---")
print(df.describe())
```
